```python
import matplotlib.pyplot as plt 
```

```python
data["Gender"].value_counts().sort_values(ascending= False).plot(kind="pie")
plt.title("Pie Chart For Gender")
plt.ylabel("")
plt.show()
```

```python
data.head(2)
```

```python
plt.hist(data["Job Rate"])
plt.title("Job Rates")
plt.xlabel("Rate")
plt.ylabel("Count")
plt.show()
```

```python
data["Job Rate"].describe()
```

```python
data.groupby("Department")["Annual Salary"].mean().sort_values(ascending= False).head(7).plot(kind="bar")
plt.title("Avrage Salary (By department)")
plt.xlabel("Department")
plt.ylabel("Salary")
plt.show()
```
